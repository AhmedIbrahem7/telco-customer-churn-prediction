* The analysis will investigate customer demographics, service usage, contract type, payment methods, tenure, and charges to identify patterns and relationships with churn.

* What is the overall churn rate?
* Which customer characteristics are associated with churn?
* Does customer tenure affect churn?
* Does the type of contract affect churn?
* Do monthly and total charges differ between churned and retained customers?
* Which services are associated with higher or lower churn?

* ### Feature Engineering

```python
# Converting the target variable Churn into numerical values

df['Churn'] = df['Churn'].map({
    'No': 0,
    'Yes': 1
})
```

```python
df['Churn'].value_counts()
```

```python
# Creating a new feature representing the total number of subscribed services

service_cols = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

df['TotalServices'] = (df[service_cols] == 'Yes').sum(axis=1)

df[['TotalServices', 'Churn']].head()
```

```python
# Creating a feature representing the average monthly spend

df['AverageMonthlySpend'] = np.where(
    df['tenure'] > 0,
    df['TotalCharges'] / df['tenure'],
    df['MonthlyCharges']
)

df[['tenure', 'MonthlyCharges', 'TotalCharges', 'AverageMonthlySpend']].head()
```

```python
# Removing the feature created only for exploratory analysis

df = df.drop(columns=['TenureGroup'])
```

* ### Data Preprocessing

```python
# Separating the features from the target variable

X = df.drop(columns=['Churn'])
y = df['Churn']
```

```python
# Displaying the selected features

print("Number of features:", X.shape[1])
print("\nSelected features:")
print(X.columns.tolist())
```

```python
# Identifying numerical and categorical features

numerical_features = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = X.select_dtypes(include=['str']).columns.tolist()

print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
# Splitting the dataset into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```

```python
# Creating preprocessing steps for numerical and categorical features

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numerical_features
        ),
        (
            'cat',
            OneHotEncoder(handle_unknown='ignore'),
            categorical_features
        )
    ]
)
```

```python
# Creating a pipeline combining preprocessing and Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

logistic_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LogisticRegression(max_iter=1000, random_state=42))
    ]
)

# Training the Logistic Regression model

logistic_model.fit(X_train, y_train)
```

```python
# Training the Logistic Regression model

logistic_model.fit(X_train, y_train)
```

```python
# Making predictions on the test set

y_pred_logistic = logistic_model.predict(X_test)
```

```python
# Evaluating Logistic Regression performance

logistic_results = {
    'Model': 'Logistic Regression',
    'Accuracy': accuracy_score(y_test, y_pred_logistic),
    'Precision': precision_score(y_test, y_pred_logistic),
    'Recall': recall_score(y_test, y_pred_logistic),
    'F1 Score': f1_score(y_test, y_pred_logistic)
}

logistic_results
```

```python
# Creating a pipeline combining preprocessing and Random Forest

from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight='balanced'
        ))
    ]
)
```

```python
# Training the Random Forest model

random_forest_model.fit(X_train, y_train)
```

```python
# Making predictions on the test set

y_pred_rf = random_forest_model.predict(X_test)
```

```python
# Evaluating Random Forest performance

random_forest_results = {
    'Model': 'Random Forest',
    'Accuracy': accuracy_score(y_test, y_pred_rf),
    'Precision': precision_score(y_test, y_pred_rf),
    'Recall': recall_score(y_test, y_pred_rf),
    'F1 Score': f1_score(y_test, y_pred_rf)
}

random_forest_results
```

```python
# Creating a pipeline combining preprocessing and Gradient Boosting

from sklearn.ensemble import GradientBoostingClassifier

gradient_boosting_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', GradientBoostingClassifier(
            random_state=42
        ))
    ]
)
```

```python
# Training the Gradient Boosting model

gradient_boosting_model.fit(X_train, y_train)
```

```python
# Making predictions on the test set

y_pred_gb = gradient_boosting_model.predict(X_test)
```

```python
# Evaluating Gradient Boosting performance

gradient_boosting_results = {
    'Model': 'Gradient Boosting',
    'Accuracy': accuracy_score(y_test, y_pred_gb),
    'Precision': precision_score(y_test, y_pred_gb),
    'Recall': recall_score(y_test, y_pred_gb),
    'F1 Score': f1_score(y_test, y_pred_gb)
}

gradient_boosting_results
```

```python
# Comparing the performance of the three models

model_results = pd.DataFrame([
    logistic_results,
    random_forest_results,
    gradient_boosting_results
])

model_results.sort_values(
    by='F1 Score',
    ascending=False
)
```

```python
# Defining hyperparameters for Logistic Regression tuning

from sklearn.model_selection import GridSearchCV

param_grid = {
    'model__C': [0.01, 0.1, 1, 10, 100]}
```

```python
# Defining hyperparameters for Logistic Regression tuning

param_grid = {
    'model__C': [0.01, 0.1, 1, 10, 100]}
```

```python
# Creating GridSearchCV for Logistic Regression

grid_search = GridSearchCV(
    estimator=logistic_model,
    param_grid=param_grid,
    scoring='recall',
    cv=5,
    n_jobs=-1
)
```

```python
# Training GridSearchCV

grid_search.fit(X_train, y_train)
```

```python
# Displaying the best hyperparameter

print("Best parameter:", grid_search.best_params_)
print("Best cross-validation recall:", grid_search.best_score_)
```

```python
# Making predictions using the tuned Logistic Regression model

best_logistic_model = grid_search.best_estimator_

y_pred_tuned = best_logistic_model.predict(X_test)
```

```python
# Evaluating the tuned Logistic Regression model

tuned_logistic_results = {
    'Model': 'Tuned Logistic Regression',
    'Accuracy': accuracy_score(y_test, y_pred_tuned),
    'Precision': precision_score(y_test, y_pred_tuned),
    'Recall': recall_score(y_test, y_pred_tuned),
    'F1 Score': f1_score(y_test, y_pred_tuned)
}

tuned_logistic_results
```

```python
# Comparing the original and tuned Logistic Regression models

final_comparison = pd.DataFrame([
    logistic_results,
    random_forest_results,
    gradient_boosting_results,
    tuned_logistic_results
])

final_comparison.sort_values(
    by='F1 Score',
    ascending=False
)
```

```python
# Creating the confusion matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_tuned)

cm
```

```python
# Visualizing the confusion matrix

fig = px.imshow(
    cm,
    text_auto=True,
    x=['Predicted No Churn', 'Predicted Churn'],
    y=['Actual No Churn', 'Actual Churn'],
    title='Confusion Matrix - Tuned Logistic Regression',
    labels={
        'x': 'Predicted',
        'y': 'Actual',
        'color': 'Count'
    }
)

fig.show()
```

```python
# Generating the classification report

from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred_tuned,
    target_names=['No Churn', 'Churn']
))
```

```python
# Validating the final model using 5-fold cross-validation

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    best_logistic_model,
    X_train,
    y_train,
    cv=5,
    scoring='recall'
)

print("Cross-validation recall scores:", cv_scores)
print("Mean cross-validation recall:", cv_scores.mean())
```

```python
# Creating a final evaluation summary

final_metrics = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1 Score'],
    'Score': [
        accuracy_score(y_test, y_pred_tuned),
        precision_score(y_test, y_pred_tuned),
        recall_score(y_test, y_pred_tuned),
        f1_score(y_test, y_pred_tuned)
    ]
})

final_metrics
```
